## Checking Routes Between Cities Using DFS

This notebook demonstrates how to use Depth-First Search (DFS) to analyze a road network represented as a graph.

### 1. Representing the Road Network as a Graph

We'll use an adjacency list to represent the graph, where each city is a key in a dictionary, and its value is a list of cities it's connected to.

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
    'G': ['H'],
    'H': ['G']
}

print("Road Network Graph:")
for city, connections in graph.items():
    print(f"{city}: {connections}")
```

### 2. Implementing DFS to Check for a Route Between Two Cities

This function will use DFS to determine if there is a path from a `start` city to an `end` city.

```python
def has_route_dfs(graph, start, end, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)

    if start == end:
        return True

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            if has_route_dfs(graph, neighbor, end, visited):
                return True
    return False

# Test cases
print("\n--- Route Existence Check ---")
print(f"Route from A to E: {has_route_dfs(graph, 'A', 'E')}")
print(f"Route from A to G: {has_route_dfs(graph, 'A', 'G')}")
print(f"Route from G to H: {has_route_dfs(graph, 'G', 'H')}")
print(f"Route from Z to A (non-existent city): {has_route_dfs(graph, 'Z', 'A')}")
```

### 3. Implementing DFS to Find All Connected Components

This function will iterate through all cities and use DFS to explore each connected component, ensuring each city is visited only once.

```python
def find_connected_components_dfs(graph):
    visited = set()
    components = []

    def dfs_explore(node, current_component):
        visited.add(node)
        current_component.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                dfs_explore(neighbor, current_component)

    for city in graph:
        if city not in visited:
            current_component = []
            dfs_explore(city, current_component)
            components.append(current_component)

    return components

# Find and display connected components
print("\n--- Connected Components ---")
connected_components = find_connected_components_dfs(graph)
for i, component in enumerate(connected_components):
    print(f"Component {i+1}: {component}")
```
