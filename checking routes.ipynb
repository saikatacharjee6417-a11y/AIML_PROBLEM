Checking Routes Between Cities Using DFS:

A small road network connects several cities. Represent the road network as a graph, where cities are vertices and roads are edges.

Write a Python program using Depth-First Search (DFS) to:

Check whether a route exists between two given cities.

Find and display all connected components of the graph, where each connected component is a group of cities that are connected to each other but not connected to any city in another group.



```python
# Checking Routes Between Cities Using DFS

# Define the cities and roads
graph = {
    "Kolkata": ["Delhi", "Bhubaneswar"],
    "Delhi": ["Kolkata", "Mumbai", "Jaipur"],
    "Mumbai": ["Delhi", "Pune"],
    "Pune": ["Mumbai"],
    "Jaipur": ["Delhi"],
    "Bhubaneswar": ["Kolkata"],
    "Chennai": ["Bengaluru"],
    "Bengaluru": ["Chennai", "Hyderabad"],
    "Hyderabad": ["Bengaluru"]
}


# Function to check whether a route exists
def route_exists(graph, start, destination):
    visited = set()

    def dfs(city):
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph[city]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True

        return False

    return dfs(start)


# Function to find all connected components
def find_components(graph):
    visited = set()
    components = []

    def dfs(city, component):
        visited.add(city)
        component.append(city)

        for neighbour in graph[city]:
            if neighbour not in visited:
                dfs(neighbour, component)

    for city in graph:
        if city not in visited:
            component = []
            dfs(city, component)
            components.append(component)

    return components


# Given two cities
start_city = "Kolkata"
destination_city = "Pune"

# Check route
if route_exists(graph, start_city, destination_city):
    print(f"Route exists between {start_city} and {destination_city}.")
else:
    print(f"No route exists between {start_city} and {destination_city}.")


# Find connected components
components = find_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
