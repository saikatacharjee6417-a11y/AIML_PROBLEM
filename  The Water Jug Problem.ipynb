## Problem 2: The Water Jug Problem (Solved using BFS)

**Problem Statement:**
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque

def water_jug_problem(jug_a_capacity, jug_b_capacity, target):
    """
    Solves the Water Jug Problem using Breadth-First Search (BFS).

    Args:
        jug_a_capacity (int): Capacity of Jug A.
        jug_b_capacity (int): Capacity of Jug B.
        target (int): The target amount of water to measure.

    Returns:
        list: A list of steps to reach the target, or None if not possible.
    """

    # States are represented as tuples (water_in_a, water_in_b)
    initial_state = (0, 0)
    queue = deque([(initial_state, [])])  # (current_state, path_taken)
    visited = {initial_state}

    while queue:
        (current_a, current_b), path = queue.popleft()

        # Check if target is reached
        if current_a == target or current_b == target:
            return path + [(current_a, current_b)]

        # Possible operations
        next_states = []

        # 1. Fill Jug A
        next_states.append((jug_a_capacity, current_b, "Fill Jug A"))
        # 2. Fill Jug B
        next_states.append((current_a, jug_b_capacity, "Fill Jug B"))
        # 3. Empty Jug A
        next_states.append((0, current_b, "Empty Jug A"))
        # 4. Empty Jug B
        next_states.append((current_a, 0, "Empty Jug B"))

        # 5. Pour A to B
        pour_a_to_b = min(current_a, jug_b_capacity - current_b)
        next_states.append((current_a - pour_a_to_b, current_b + pour_a_to_b, "Pour A to B"))

        # 6. Pour B to A
        pour_b_to_a = min(current_b, jug_a_capacity - current_a)
        next_states.append((current_a + pour_b_to_a, current_b - pour_b_to_a, "Pour B to A"))

        for next_a, next_b, action in next_states:
            if (next_a, next_b) not in visited:
                visited.add((next_a, next_b))
                queue.append(((next_a, next_b), path + [(current_a, current_b, action)]))
    return None

# Define jug capacities and target
JUG_A_CAP = 4
JUG_B_CAP = 3
TARGET = 2

# Solve the problem
solution_path = water_jug_problem(JUG_A_CAP, JUG_B_CAP, TARGET)

# Print the solution
if solution_path:
    print(f"Minimum steps to measure {TARGET} litres:")
    for i, (a, b, *action) in enumerate(solution_path):
        step_description = f"Step {i}: {action[0] if action else 'Initial State'}"
        print(f"{step_description.ljust(30)} -> Jug A: {a}L, Jug B: {b}L")
else:
    print(f"Target {TARGET} litres cannot be reached with Jug A capacity {JUG_A_CAP}L and Jug B capacity {JUG_B_CAP}L.")
```
